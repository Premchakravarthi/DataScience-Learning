- the input is getting change every time, and the rest of the code is same, and this is like how many type of ways you can give the input.

```python
user_distance = eval(input('enter the user distance'))
govt_distance = eval(input('enter the govt distance'))   #50
if user_distance > govt_distance:
    charge_amount = eval(input('enter the charge amount value:'))
    charge = (user_distance - govt_distance) * charge_amount
    update = user_distance - govt_distance
    print(f'chargable distance {update}')
    print(f'Total charge payable {charge}rs')
else:
    print('Enjoy your free ride')
```

```python
import random
user_distance = random.randint(1,100)
govt_distance = eval(input('enter the govt distance'))
if user_distance > govt_distance:
    charge_amount = eval(input('enter the charge value:'))
    update = user_distance - govt_distance
    print(f'you need to pay charge for {update}km')
    charge = (user_distance - govt_distance) * charge_amount
    print(f'Total amount you need to pay for extra distance {charge}')
else:
    print('enjoy your free ride')
```
