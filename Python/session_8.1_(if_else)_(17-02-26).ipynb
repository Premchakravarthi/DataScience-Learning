```python
n1 = eval(input("enter number 1:"))
n2 = eval(input("enter number 2:"))
operation = eval(input("choose operation:"))
if operation == 1:
    add = n1 + n2
    print(f"addition of {n1} and {n2} :", add)
elif operation == 2:
    sub = n1 - n2
    print(f"subtraction of {n1} and {n2} :",sub)
elif operation == 3:
    mul = n1 * n2
    print(f"multiplication of {n1} and {n2}:",mul)
elif operation == 4:
    div = n1 / n2
    print(f"division of {n1} and {n2} :",div)
else:
    print("Enter a valid operation number, try again!")
```
