- Variables are case sensitive

    - which means both capital and small letters are different variables
    - Ex: NUMBER=100, number=200
- Variables are allowed number as suffix

    - Ex: number123=300
- Variables are not allowed number as prefix

    - Ex:123number=400
- Variables does not allow special charcters

- Only underscore can be allowed : number_one=1000

- Keywords are not allowed to use variables

- 35-36 keywords are there like if,else,for,while etc
- In built functions can be allowed, but not recommended
