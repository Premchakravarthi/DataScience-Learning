```python
# WAP take three numbers, find the average
n1 = eval(input('Enter a number:'))
n2 = eval(input('Enter a number:'))
n3 = eval(input('Enter a number:'))
avg = (n1 + n2 + n3)/3
print(f'The average of {n1}, {n2} and {n3} is {avg}.')
```

```python
# WAP calculate area of circle when radius given as 20
R = eval(input('Enter Radius:'))
pi = eval(input('Enter pi value:'))
area_of_circle = pi * R * R
print(f'The area of the circle is {area_of_circle}.')
```

```python
# WAP calculate area of right angle traingle when
# height given as 20, breadth given as 50
# Formual= 1/2*breadth*height
H = eval(input('Enter Height Value:'))
B = eval(input('Enter Breadth Value:'))
area_of_right_angle_triangle = 1/2 * B * H
print(f'The area of the right angle triangle is {area_of_right_angle_triangle}.')
```

```python
# Wap Calculate area of rectangle
L = eval(input('Enter Length value:'))
B = eval(input('Enter Breadth value:'))
Area_of_rectangle = L * B
print(f'The Area of the rectangle is {Area_of_rectangle}.')
```

```python
# Wap calculate the weight in kgs when weight in pounds
Pounds = eval(input('Enter weight value in pounds:'))
Weight_in_kgs = Pounds * 2.2
print(f'The weight of pounds in Kgs is {Weight_in_kgs}.')
```

```python
# Wap calculate the money in rupees when money in dollars
dollars = eval(input('Enter amount in dollars:'))
rupees = dollars * 90
print(f'The total amount in rupees is {rupees}rs.')
```

```python
# Wap ask the user to calculate the tax amount for a given salary of 100000, and the tax percentage is 10
salary = 100000
tax_percentage = 10
tax_amount = (salary * tax_percentage) / 100
print(f'Salary: {salary}')
print(f'Tax Percentage: {tax_percentage}%')
print(f'Total Tax Amount: {tax_amount}rs')
```
